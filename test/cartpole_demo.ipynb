```python
import numpy as np, matplotlib.pyplot as plt
from cartpole_wind_pch import CartPoleWindPCH
from ucbvi import UCBVI
```

```python
# 1. Setup
env   = CartPoleWindPCH(wind_std=0.05)
agent = UCBVI(bins_per_dim=6, n_actions=env.env.action_space.n,
              horizon=env.env.max_episode_steps)

returns = []

# 2. Training loop
for ep in range(200):
    obs, _ = env.env.reset()
    s      = agent.discretize(obs)
    agent.plan()                     # optimistic Q‑init
    ep_ret = 0

    for t in range(agent.H):
        a  = agent.act(s)
        o2, r, term, trunc, _ = env.do(a)
        s2 = agent.discretize(o2)
        agent.update(s, a, r, s2)
        s  = s2
        ep_ret += r
        if term or trunc:
            break

    returns.append(ep_ret)

# 3. Visualise
plt.plot(returns)
plt.xlabel('episode'); plt.ylabel('return'); plt.show()
```
