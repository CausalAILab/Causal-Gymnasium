```python
from causal_gym.envs.race import RacePCH
```

```python
num_steps = 200
pch = RacePCH(num_steps=num_steps, render_mode='human', perception='truth')

pch.reset()

for t in range(num_steps):
    action, obs, reward, terminated, truncated, info = pch.see(show_reward=True)
    pch.render()

    if terminated or truncated:
        pch.close()
        break

print(pch.env._D)
print(pch.env.X)
print(pch.env.H)
print(pch.env.C)
print(pch.env.W)
print(pch.env._U)
print(pch.env._Y)
print(pch.env._Y[-1])
```
