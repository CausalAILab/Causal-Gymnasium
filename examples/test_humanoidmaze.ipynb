```python
import ogbench
import numpy as np

env = ogbench.make_env_and_datasets(
    'humanoidmaze-medium-navigate-singletask-task3-v0',
    env_only=True,
)

print('Observation space:', env.observation_space)
print('Action space:    ', env.action_space)

obs, info = env.reset(options=dict(task_id=1))
print(f'obs shape: {obs.shape}')
print(f'obs values: {obs}')
print()
print('goal shape:', info['goal'].shape)
print('goal values:', info['goal'])

# Check the underlying MuJoCo state
unwrapped = env.unwrapped
qpos = unwrapped.data.qpos.copy()
qvel = unwrapped.data.qvel.copy()
print(f'\nqpos (nq={len(qpos)}): {qpos}')
print(f'qvel (nv={len(qvel)}): {qvel}')

print('\n--- Comparing obs dims to qpos/qvel ---')
for i, v in enumerate(obs):
    print(f'  obs[{i}] = {v:.6f}')
```
